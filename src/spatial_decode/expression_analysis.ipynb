# Estimate cell-type signatures
The following description is taken from the Project description - Analysis track (02_analysis_track_spec.pdf).

*"From reference/— not from ground_truth/. For each cell type, average the reference cells
of that type to get a mean expression profile. Stack them into a signature matrix S with genes
as rows and cell types as columns.
Because the reference is a finite sample, S is an estimate. How good it is depends on how many
reference cells you were given — which is one of the difficulty knobs."*

```python
# Functions to import reference_counts.csv and reference_labels.csv
import pandas as pd

def load_reference_counts(path_ref_counts):
    ref_counts = pd.read_csv(path_ref_counts)
    return ref_counts
    
def load_reference_labels(path_ref_labels):
    ref_labels = pd.read_csv(path_ref_labels)
    return ref_labels
```

```python
# Function to merge the two data sets by cell_id 

def merge_counts_types(ref_counts, ref_labels):
    merged = ref_counts.merge(ref_labels, on="cell_id", how="inner")
    return merged
    
# Function to generate the matrix S 
def mean_expression_matrix(merged):
    S = merged.groupby("cell_type").mean(numeric_only=True).T
    return S
```

```python
# Function to write results in csv

def write_matrix(S, results_path):
    S.to_csv(results_path)
```

```python
write_matrix(mean_expression_matrix(merge_counts_types(load_reference_counts("/Users/janine/pls2026-analysis/data/raw/reference_counts.csv"),
                                                       load_reference_labels("/Users/janine/pls2026-analysis/data/raw/reference_labels.csv"))),
            "/Users/janine/pls2026-analysis/results_spatial_decode/mean_expression_profile_reference.csv")
```
